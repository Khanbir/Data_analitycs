# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.

**Формулювання гіпотез і типу тесту**
Нульова гіпотеза (H0): Середній час до покупки не збільшився. Це означає, що середній час після редизайну дорівнює або менше історичного значення.

(H0) μ =< 123.7

Альтернативна гіпотеза (Ha): Середній час до покупки збільшився. Це означає, що середній час після редизайну більший за історичне значення.

(Ha) μ>123.7

Тип тесту: Оскільки ми перевіряємо, чи середній час збільшився, це правосторонній тест.

2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


```python
## 2. Обчисліть вибіркові статистики:  x¯ ,  s ,  n .
import numpy as np
from scipy import stats

# Задані параметри
historical_mean = 123.7
sample_data = np.array([128, 135, 121, 142, 126, 151, 123])
alpha_1 = 0.10
alpha_2 = 0.05

# Вибіркові статистики
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)
sample_size = len(sample_data)

print(f"Вибіркове середнє (x̄): {sample_mean:.2f}")
print(f"Вибіркове стандартне відхилення (s): {sample_std:.2f}")
print(f"Розмір вибірки (n): {sample_size}");

## 3
# Виконуємо односторонній t-тест (right-tailed)
t_statistic, p_value = stats.ttest_ind(
    a=sample_data,
    b=np.full(len(sample_data), historical_mean),
    alternative='greater',
    equal_var=True
)
print("\n--- Результати T-тесту ---")
print(f"T-статистика: {t_statistic:.3f}")
print(f"P-value: {p_value:.3f}")

# Для порівняння: критичне t-значення для alpha_1 = 0.10
degrees_of_freedom = sample_size - 1
critical_t_1 = stats.t.ppf(1 - alpha_1, df=degrees_of_freedom)
print(f"Критичне t-значення для α=0.10: {critical_t_1:.3f}")
```

4. Прийняття рішення та    5. Висновок (за α=0.10)
Порівняння p-value з
alpha=0.10: p_value (0.043) < alpha_1 (0.10).

**Рішення**: Оскільки p-value менше рівня значущості, ми відхиляємо нульову гіпотезу.

**Висновок**: На рівні значущості
alpha=0.10, є достатньо доказів, щоб стверджувати, що середній час до покупки після редизайну сайту збільшився.

6. Зміна рівня значущості на

Порівняння p-value з

**Рішення**: Оскільки p-value все ще менше нового рівня значущості, наше рішення не зміниться. Ми продовжуємо відхиляти нульову гіпотезу.

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

```python
import statsmodels.stats.proportion as smp

# Параметри задачі
count = 311  # Кількість успіхів у вибірці
nobs = 1200  # Розмір вибірки
value = 0.23  # Пропорція в популяції
alpha = 0.03  # Рівень значущості

# Виконання Z-тесту для пропорцій
z_statistic, p_value = smp.proportions_ztest(count, nobs, value=value, alternative='larger')

print(f"Z-статистика: {z_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Висновок за p-value
if p_value < alpha:
    print(f"\nВисновок: P-value ({p_value:.4f}) < α ({alpha}), тому ми відхиляємо нульову гіпотезу.")
    print("Є достатньо доказів, щоб стверджувати, що впізнаваність логотипу зросла.")
else:
    print(f"\nВисновок: P-value ({p_value:.4f}) > α ({alpha}), тому ми не відхиляємо нульову гіпотезу.")
    print("Немає достатньо доказів, щоб стверджувати, що впізнаваність логотипу зросла.")
```

Z-статистика: 2.3708
P-value: 0.0089

Висновок: Оскільки p-value (0.0089) менше за рівень значущості
alpha (0.03), ми відхиляємо нульову гіпотезу. Це дає нам достатні підстави стверджувати, що після рекламної кампанії впізнаваність логотипу зросла і тепер становить більше ніж 23%.

```python
## 2. Обчислення та інтерпретація довірчого інтервалу
# Обчислення довірчого інтервалу
confidence_interval = smp.proportion_confint(count, nobs, alpha=alpha, method='normal')

print(f"\nДовірчий інтервал (97%): ({confidence_interval[0]:.4f}, {confidence_interval[1]:.4f})")
```

Інтерпретація:
Ми можемо з 97% впевненістю стверджувати, що справжня частка дорослих, які впізнають логотип, знаходиться в діапазоні від 23.74% до 28.10%.

Як він пояснює прийняте рішення?
Отриманий довірчий інтервал (0.2374, 0.2810) повністю лежить вище початкової пропорції популяції 23% (0.23). Це означає, що значення 23% знаходиться поза цим інтервалом. Цей факт підтверджує наш висновок, зроблений на основі p-value та Z-статистики: істинна пропорція популяції, ймовірно, вища за 23%, що свідчить про успіх рекламної кампанії.
