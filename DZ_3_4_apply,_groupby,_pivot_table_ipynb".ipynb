3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoices_by_category = df.pivot_table(
    index='Branch',
    columns='Product line',
    values='Invoice ID',
    aggfunc='count',
    )
invoices_by_category
```

3.2. Викристовуючи змінну `invoices_by_category` дайте відповідь програмно (тобто значення треба не просто знайти очима, а вивести за допомогою коду), в якому філіалі магазину (`Branch`) найбільше інвойсів із покупкою товарів категорії "Електронні аксесуари" (`Electronic accessories`)?


```python
max_branch = invoices_by_category['Electronic accessories'].idxmax()
max_value = invoices_by_category['Electronic accessories'].max()

print(f'Найбільше інвойсів для Electronic accessories у філії {max_branch} — {max_value} покупок.')
```

Для отримання більше відповідей щодо стану справ у підпорядкованих філіалах хотілося отримати відповіді:
1. в яких магазинах (містах) які товари користуються більшим рейтингом?
2. в яких магазинах меншим рейтингом?
3. в які дні пікове навантаження?


```python
invoices_by_city_cat = df.pivot_table(
    index='City',
    columns='Product line',
    values='Rating',
    aggfunc='mean',
    )
invoices_by_city_cat

max_city = invoices_by_city_cat.idxmax()
min_city = invoices_by_city_cat.idxmin()
display ('Товари з максимальними рейтингами по містах', max_city, 'Товари з мінімальними рейтингами по містах', min_city)
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Weekday'] = df['Date'].dt.day_name()

low_sales_by_branch = df.groupby(['City', 'Weekday'])['Total'].sum().reset_index()
least_busy_day = low_sales_by_branch.sort_values(['City', 'Total']).groupby('City').first()
least_busy_day
```
